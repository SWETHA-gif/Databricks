🎯 Module 3 – Product Ranking

Rank products by:

Revenue,  
Profit,  
Quantity Sold

```python
%sql

SELECT
    `Product ID`,
    `Product Name`,
    ROUND(SUM(`Sales`),2) AS Total_Sales,
    ROUND(SUM(`Profit`),2) AS Total_Profit,
    SUM(`Quantity`) AS Total_Quantity,

    RANK() OVER(
        ORDER BY SUM(`Sales`) DESC
    ) AS Sales_Rank,

    RANK() OVER(
        ORDER BY SUM(`Profit`) DESC
    ) AS Profit_Rank,

    RANK() OVER(
        ORDER BY SUM(`Quantity`) DESC
    ) AS Quantity_Rank

FROM abc_retail_pvt_ltd.data_analytics.superstore

GROUP BY
    `Product ID`,
    `Product Name`;
```

Business asks:

"How much revenue has accumulated up to each month?"

```python
%sql
WITH MonthlySales AS 
(
SELECT YEAR(`Order Date`) AS Year, MONTH(`Order Date`) AS Month, ROUND(SUM(Sales),2) AS Monthly_Sales
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY
YEAR(`Order Date`),
MONTH(`Order Date`)
)

SELECT * FROM MonthlySales;
```

```python
%sql
WITH MonthlySales AS
(
SELECT YEAR(`Order Date`) AS Year, MONTH(`Order Date`) AS Month, ROUND(SUM(Sales),2) AS Monthly_Sales
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY
YEAR(`Order Date`),
MONTH(`Order Date`)
)

SELECT Year, Month, Monthly_Sales, SUM(Monthly_Sales) OVER(ORDER BY Year, Month) AS Running_Total
FROM MonthlySales;
```

Why no PARTITION BY?

Because business asked:

Total revenue accumulated over time.

We don't want the total to restart.

If we wrote

PARTITION BY Year

Then

Running total becomes

2023 → restart

2024 → restart

2025 → restart

Different business question.

Find the second highest selling customer.

```python
%sql
WITH CustomerSales AS 
(
SELECT `Customer Name`, SUM(`Sales`) AS Total_Sales, RANK() OVER(ORDER BY SUM(`Sales`) DESC) AS Customer_Rank
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY `Customer Name`
)

SELECT * FROM CustomerSales
WHERE Customer_Rank = 2;
```

What problem does a Window Function solve that GROUP BY cannot?

Window functions allow us to perform calculations such as ranking, running totals & comparisons across rows while retaining the original rows. Unlike GROUP BY, they do not collapse multiple rows into a single aggregated row.
