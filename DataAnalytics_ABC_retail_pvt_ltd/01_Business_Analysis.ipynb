Check for NULL Values

A Data Analyst always checks data quality before doing any analysis.

```python
%sql
-- Check for NULL Values
-- A Data Analyst always checks data quality before doing any analysis.
SELECT
SUM(CASE WHEN `Sales` IS NULL THEN 1 ELSE 0 END) AS sales_nulls,
SUM(CASE WHEN `Profit` IS NULL THEN 1 ELSE 0 END) AS profit_nulls,
SUM(CASE WHEN `Customer ID` IS NULL THEN 1 ELSE 0 END) AS customer_nulls,
SUM(CASE WHEN `Product ID` IS NULL THEN 1 ELSE 0 END) AS product_nulls
FROM abc_retail_pvt_ltd.data_analytics.superstore;
```

```python
%sql
-- Check for Duplicate Rows
select *, count(*) as duplicate_count from abc_retail_pvt_ltd.data_analytics.superstore
group by all
having count(*) > 1;
```

📊 Sales Analysis

What is the total revenue? 

Which month has the highest sales? 

Which region performs best?

```python
%sql
-- Total Revenue
select sum(`Sales`) as Total_Revenue
from abc_retail_pvt_ltd.data_analytics.superstore; 
```

```python
%sql
-- Which month has the highest sales?
SELECT
YEAR(`Order Date`) AS Year,
MONTH(`Order Date`) AS Month,
SUM(Sales) AS TotalSales
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY
YEAR(`Order Date`),
MONTH(`Order Date`)
ORDER BY TotalSales DESC
LIMIT 1;
```

```python
%sql
-- Which region performs best?
select sum(`Sales`) as Total_Sales_Region, `Region`
from abc_retail_pvt_ltd.data_analytics.superstore
group by Region
order by Total_Sales_Region desc;
```

👥 Customer Analysis

Who are the top 10 customers?

Who generates the most profit?

Which segment is most valuable?

```python
%sql
-- Who are the top 10 customers?
select `Customer Name`, sum(`Sales`) as Total_Sales
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer Name`
order by Total_Sales desc
limit 10;
```

```python
%sql
-- Who generates the most profit?
select `Customer Name`, sum(`Profit`) as Total_Profit
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer Name`
order by Total_Profit desc
limit 10;
```

```python
%sql
-- Which segment is most valuable?
SELECT
Segment,
SUM(Sales) Sales,
SUM(Profit) Profit
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY Segment;
```

📦 Product Analysis

Best-selling products

Loss-making products

High-discount, low-profit products


```python
%sql
-- Best-selling products
select sum(`Quantity`), `Product Name`
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Product Name`
order by sum(`Quantity`) desc
limit 10
```

```python
%sql
-- Loss-making products
select `Product Name`, sum(`Profit`) as Total_Profit
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Product Name`
order by Total_Profit asc
limit 10
```

```python
%sql
-- High-discount, low-profit products
SELECT
`Product Name`,
AVG(Discount) AS AvgDiscount,
SUM(Profit) AS TotalProfit
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY `Product Name`
ORDER BY TotalProfit ASC, AvgDiscount DESC;
```
