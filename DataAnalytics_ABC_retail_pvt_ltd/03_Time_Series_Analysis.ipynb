Task 1

Find the first and last order date.

Think:

```python
%sql
SELECT
    MIN(`Order Date`) AS Earliest_Order_Date,
    MAX(`Order Date`) AS Latest_Order_Date
FROM abc_retail_pvt_ltd.data_analytics.superstore;
```

The year from 2023 to 2026 so 4 years of data were available to be analyzed and know the business status

The dataset contains four years of historical sales data (2023–2026), providing sufficient information to analyze sales trends, customer behavior, and business performance over time.

Task 2

How many orders were placed each year?

Business Question:

Is the company growing every year?

```python
%sql
select year(`Order Date`), count(distinct `Order ID`) from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`)
order by year(`Order Date`);
```

The number of customer orders increased consistently each year, growing from 2051 orders in 2023 to 3379 orders in 2026. This indicates positive business growth and increased customer demand.

```python
%sql
select year(`Order Date`), Round ((sum(`Profit`)/sum(`Sales`))*100,2) as Profit_margin 
from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`)
order by year(`Order Date`);
```

Profit margin slightly declined from 13.46% to 12.87% in 2026, suggesting that although sales increased, profitability did not improve at the same rate.

Task 3

Revenue by Year

Business Question:

Which year generated the highest revenue?

```python
%sql
select year(`Order Date`), round(sum(`Sales`),2) as Total_Revenue 
from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`)
order by total_revenue desc;
```

Revenue increased year over year, with 2026 recording the highest annual revenue, indicating sustained business growth.

Task 4

Profit by Year

Business Question:

Which year generated the highest profit?

```python
%sql
select year(`Order Date`), round(sum(`Profit`),2) as Total_Profit 
from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`)
order by total_profit desc;
```

2026 generated the highest annual profit of $95,926, indicating improved financial performance compared to previous years.

📌 Part 2 – Monthly Analysis

Now the Sales Director asks:

Task 5

Monthly Revenue

⚠️ Remember our previous discussion.

Don't group only by month.

Think carefully.

```python
%sql
select year(`Order Date`), Month(`Order Date`), Round(sum(`Sales`),2) as Monthy_Profile
from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`), Month(`Order Date`)
order by year(`Order Date`), Month(`Order Date`);
```

Finding

Sales consistently increase during November and December.

Possible Reason

Holiday shopping and year-end promotions.

Recommendation

Increase inventory and marketing before Q4.

Task 7

Monthly Orders

```python
%sql
select year(`Order Date`), Month(`Order Date`), count(`Order ID`) as Monthy_orders
from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`), Month(`Order Date`)
order by year(`Order Date`), Month(`Order Date`);
```

Task 8

Monthly Average Order Value

```python
%sql
select year(`Order Date`), Month(`Order Date`), round(sum(`Sales`) / count(distinct `Order ID`),2) as AOV 
from abc_retail_pvt_ltd.data_analytics.superstore
group by year(`Order Date`), Month(`Order Date`)
order by year(`Order Date`), Month(`Order Date`);
```

📌 Part 3 – Quarterly Analysis

Business Question:

Which quarter performs best?

Find:

Revenue
Profit
Orders

for every quarter.

```python
%sql
SELECT
    YEAR(`Order Date`) AS Year,
    CONCAT('Q', QUARTER(`Order Date`)) AS Quarter,
    ROUND(SUM(`Sales`), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit,
    COUNT(DISTINCT `Order ID`) AS Orders
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY YEAR(`Order Date`), QUARTER(`Order Date`)
ORDER BY YEAR(`Order Date`), QUARTER(`Order Date`);
```

Q4 generated the highest revenue in every year.

Possible Reason

Holiday sales, Black Friday, Christmas promotions, and year-end purchasing behavior likely contributed to increased sales.

Recommendation

Allocate additional inventory, logistics, and marketing budget during Q4 to maximize revenue.

Revenue increased consistently across all four years, reaching its highest value in 2026.

This indicates sustained business growth rather than a one-time increase.

Management should continue investing in high-performing categories while analyzing the factors contributing to annual growth.
