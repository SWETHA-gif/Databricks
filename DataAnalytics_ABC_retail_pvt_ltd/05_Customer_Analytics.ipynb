📌 Module 1 – Customer Summary

For every customer, calculate:

Total Sales  
Total Profit  
Total Orders  
Total Quantity Purchased  
Average Order Value  

Expected Output

Customer	Sales	Profit	Orders	Qty	AOV

Business Insight  
Identify high-value customers.  
Measure customer profitability.  
Compare purchasing behavior.

```python
%sql
select `Customer Name`, sum(`Sales`) as Total_Sales, sum(`Profit`) as Total_Profit, count(distinct `Order ID`) as Total_Orders, sum(`Quantity`)as Total_Quantity, 
round( sum(`Sales`) / COUNT(DISTINCT `Order ID`),2) as AOV, 
round(( sum(`Profit`) / sum(`Sales`))*100,2) as Profit_margin
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer ID`, `Customer Name`
order by total_orders desc;
```

📌 Module 2 – Customer Ranking

Rank customers based on:

Revenue
Profit
Number of Orders

Business Insight

Find:

Highest spending customers  
Most profitable customers  
Most frequent customers

```python
%sql
select `Customer Name`, `Customer ID`, round(Sum(`Sales`),2) as Total_Sales, round(Sum(Profit),2) as Total_Profit, Count(Distinct `Order ID`) as Num_Orders,
rank() over( order by Sum(`Sales`) desc) as Revenue_Rank,
rank() over(order by Sum(Profit) desc) as Profit_Rank,
rank() over(order by Count(Distinct `Order ID`) desc) as Orders_Rank
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer ID`, `Customer Name`;
```

📌 Module 3 – Customer Segmentation

This is used in almost every company.

Create segments like:

Sales	     Segment
>10000  	 Platinum
5000–10000	 Gold
1000–5000	 Silver
<1000	     Bronze

We'll use a CASE WHEN statement.

```python
%sql
select `Customer ID`, `Customer Name`, Round(sum(`Sales`),2) as Total_Sales,
CASE 
when sum(`Sales`) > 10000 then 'Platinum'
when sum(`Sales`) between 5000 and 10000 then 'Gold'
when sum(`Sales`) between 2000 and 5000 then 'Silver'
else 'Bronze'
end as Segment
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer ID`, `Customer Name`
order by Total_Sales desc;
```

📌 Module 4 – Customer Profitability

Business asks:

Which customers generate high sales but low profit?

These customers may be receiving excessive discounts.

Find them.

What should you look for?

Customers having

High Sales  
Low Profit Margin

These customers deserve investigation.

```python
%sql
select `Customer Name`, `Customer ID` , round(sum(`Sales`),2) as Total_Sales, round(sum(`Profit`),2) as Total_Profit, round((sum(`Profit`)/sum(`Sales`))*100,2) as Profit_Margin, 
rank() over( Order by sum(`Sales`) desc) as Sales_Rank
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer ID`, `Customer Name`
order by sales_rank
```

📌 Module 5 – Top 20 Customers

Business asks:

If we lose our top 20 customers, how much revenue do we lose?

Find:  

Top 20 customers  
Their revenue  
Contribution to total sales

```python
%sql
with TopCustomersRevenue as
(select `Customer ID`, `Customer Name`, round(sum(`Sales`), 2) as Total_Revenue, 
rank() over(order by sum(`Sales`) desc) as Customer_Rank
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer ID`, `Customer Name`
)

select * from TopCustomersRevenue
where Customer_Rank <= 20;
```

```python
%sql
with TopCustomersRevenue as
(
select `Customer ID`, `Customer Name`, round(sum(`Sales`), 2) as Total_Revenue, 
rank() over(order by sum(`Sales`) desc) as Customer_Rank
from abc_retail_pvt_ltd.data_analytics.superstore
group by `Customer ID`, `Customer Name`
)

select sum(Total_Revenue) from TopCustomersRevenue
where Customer_Rank <= 20;
```

```python
%sql
-- Total Revenue
select round(sum(`Sales`), 2) as Total_Revenue
from abc_retail_pvt_ltd.data_analytics.superstore
```

```python
%sql
-- Contribution of top 20 customers
WITH CustomerRanking AS
(
SELECT `Customer ID`, `Customer Name`, SUM(`Sales`) AS Total_Sales,
RANK() OVER(ORDER BY SUM(`Sales`) DESC) AS Customer_Rank
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY `Customer ID`, `Customer Name`)

SELECT ROUND((SUM(Total_Sales)/(SELECT SUM(Sales) FROM abc_retail_pvt_ltd.data_analytics.superstore))*100,2) AS Revenue_Contribution_Percentage
FROM CustomerRanking
WHERE Customer_Rank<=20;
```
