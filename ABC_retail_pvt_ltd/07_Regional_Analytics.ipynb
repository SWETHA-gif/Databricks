📌 Module 1 – Regional Performance Summary
Business Requirement

Create a regional summary.

Expected Output

Region	Sales	Profit	Orders	Quantity	Profit Margin

Think:

Total Revenue  
Total Profit  
Orders  
Quantity  
Profit Margin

```python
%sql
SELECT
    Region,
    ROUND(SUM(`Sales`),2) AS Sales,
    ROUND(SUM(`Profit`),2) AS Profit,
    COUNT(DISTINCT `Order ID`) AS Orders,
    SUM(`Quantity`) AS Quantity,
    ROUND((SUM(`Profit`) / SUM(`Sales`)) * 100,2) AS Profit_Margin
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY Region
ORDER BY Sales DESC;
```

📌 Module 2 – State Performance

Create

| State | Sales | Profit | Orders | Rank |

Business Question

Which states should receive more investment?

```python
%sql
SELECT
    `State/Province` AS State,
    ROUND(SUM(Sales),2) AS Sales,
    ROUND(SUM(Profit),2) AS Profit,
    COUNT(DISTINCT `Order ID`) AS Orders,
    DENSE_RANK() OVER (
        ORDER BY SUM(Sales) DESC
    ) AS State_Rank
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY `State/Province`
ORDER BY State_Rank;
```

📌 Module 3 – City Performance

Analyze every city.

Expected

| City | Sales | Profit | Margin |

```python
%sql
SELECT
    City,
    ROUND(SUM(Sales),2) AS Sales,
    ROUND(SUM(Profit),2) AS Profit,
    ROUND((SUM(Profit) / SUM(Sales)) * 100,2) AS Profit_Margin
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY City
ORDER BY Sales DESC;
```

📌 Module 4 – Region Ranking

Rank Regions by

Revenue  
Profit  
Orders

```python
%sql
SELECT
    Region,
    ROUND(SUM(Sales),2) AS Sales,
    ROUND(SUM(Profit),2) AS Profit,
    COUNT(DISTINCT `Order ID`) AS Orders,

    DENSE_RANK() OVER (
        ORDER BY SUM(Sales) DESC
    ) AS Revenue_Rank,

    DENSE_RANK() OVER (
        ORDER BY SUM(Profit) DESC
    ) AS Profit_Rank,

    DENSE_RANK() OVER (
        ORDER BY COUNT(DISTINCT `Order ID`) DESC
    ) AS Orders_Rank

FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY Region
ORDER BY Revenue_Rank;
```

📌 Module 5 – High Revenue, Low Profit Regions

Exactly like Product Analytics.

Business asks

Which regions sell a lot but barely make money?

```python
%sql
SELECT
    Region,
    ROUND(SUM(Sales),2) AS Sales,
    ROUND(SUM(Profit),2) AS Profit,
    ROUND((SUM(Profit) / SUM(Sales)) * 100,2) AS Profit_Margin
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY Region
ORDER BY Sales DESC, Profit ASC;
```

📌 Module 6 – Loss Making States

Business asks

Which states consistently lose money?

Hint:

Use

HAVING SUM(Profit)<0

```python
%sql
SELECT
    `State/Province` AS State,
    ROUND(SUM(Sales),2) AS Sales,
    ROUND(SUM(Profit),2) AS Profit,
    COUNT(DISTINCT `Order ID`) AS Orders
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY `State/Province`
HAVING SUM(Profit) < 0
ORDER BY Profit;
```

📌 Module 7 – Discount Analysis

This one is interesting.

Business asks

Which regions receive the highest average discount?

Expected

| Region | Avg Discount |

Business Insight

Too much discount

↓

Lower Profit

Can we reduce discounts?

```python
%sql
SELECT
    Region,
    ROUND(AVG(Discount) * 100,2) AS Avg_Discount_Percentage
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY Region
ORDER BY Avg_Discount_Percentage DESC;
```

📌 Module 8 – Regional Segmentation

Segment Regions.

Example

Sales > 700K

↓

High Performing

Sales 400K–700K

↓

Growing

Below 400K

↓

Needs Attention

```python
%sql
SELECT
    Region,
    ROUND(SUM(Sales),2) AS Sales,
    CASE
        WHEN SUM(Sales) > 700000 THEN 'High Performing'
        WHEN SUM(Sales) BETWEEN 400000 AND 700000 THEN 'Growing'
        ELSE 'Needs Attention'
    END AS Performance_Category
FROM abc_retail_pvt_ltd.data_analytics.superstore
GROUP BY Region
ORDER BY Sales DESC;
```
