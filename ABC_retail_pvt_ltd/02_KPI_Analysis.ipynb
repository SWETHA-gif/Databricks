📌 Task 6 - Average Order Value (AOV)
Business Question

On average, how much money does a customer spend in one order?

Formula:

### Total Revenue
### --------------
### Total Orders

````python
%sql
-- Average Order Value (AOV)
select round(sum(`Sales`) / count(distinct `Order ID`),2) as AOV
from abc_retail_pvt_ltd.data_analytics.superstore;
````

# 🧠 Data Analyst Thinking Process

Before writing any SQL query, always follow this approach.

```text
Business Requirement
        ↓
Understand the Business
        ↓
Write the Formula
        ↓
Think about the Dataset
        ↓
Write SQL
        ↓
Validate Results
        ↓
Business Insight
        ↓
Recommendation
```

---

## 📌 Step 1: Business Requirement
Understand what the stakeholder or business is asking.

**Example:**
> What is the Average Order Value (AOV)?

---

## 📌 Step 2: Understand the Business
Identify what the metric represents and why it is important.

**Example:**
Average Order Value tells us how much revenue is generated from a single order on average.

---

## 📌 Step 3: Write the Formula
Convert the business requirement into a mathematical formula.

**Example:**

Average Order Value = Total Sales / Total Orders

---

## 📌 Step 4: Think About the Dataset
Determine which columns are required and understand the grain of the data.

**Questions to ask:**
- Which column contains Sales?
- Which column identifies an Order?
- Does one row represent one order or one product within an order?
- Should I use `COUNT(*)` or `COUNT(DISTINCT Order ID)`?

---

## 📌 Step 5: Write SQL
Translate the formula into an SQL query.

---

## 📌 Step 6: Validate Results
Verify that the output is logical.

**Questions to ask:**
- Does the result make business sense?
- Are there any NULL values?
- Are duplicates affecting the calculation?
- Did I use the correct denominator?

---

## 📌 Step 7: Business Insight
Explain what the result means.

**Example:**
The average customer order generates **$455.20** in revenue.

---

## 📌 Step 8: Recommendation
Suggest an action based on the insight.

**Example:**
Increase the Average Order Value through cross-selling, upselling, or product bundle offers.

---

# 💡 Golden Rule

> **Never start with SQL.**
>
> Start with the business problem, understand the data, derive the formula, and then write SQL.
>
> A Data Analyst solves business problems—not just SQL queries.
